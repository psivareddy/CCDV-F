**What is Claude Reasoning Effort?**

1) Reasoning Effort controls how much internal reasoning Claude performs before generating a response. 

2) Increasing the reasoning effort allows Claude to spend more compute on planning, analysis, and solving complex problems. 

3) Higher reasoning effort generally improves performance on tasks requiring multi-step reasoning, coding, mathematics, and complex decision making.

Available Levels
Low – Fast responses for simple questions and routine tasks. 
Medium – Balanced trade-off between response quality, latency, and cost. 
High – Maximum reasoning for complex coding, architecture, and analytical problems.

```python
from  anthropic import Anthropic
claude_model="claude-haiku-4-5"
max_tokens=100
client = Anthropic()

print(f'''client : {client}, 
      \nCLAUDE_MODEL: {claude_model},
      \nmax_tokens: {max_tokens}''')
```

```python
message = client.messages.create(
    model=claude_model,
    max_tokens= max_tokens,
    messages=[
        {
            "role": "user",
            "content": "Hello, Claude",
        }
    ],
)

for block in message.content:
    if block.type == "text":
        print(block.text)
```

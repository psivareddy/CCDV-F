**Setting up the Environment**

```python
import os
from dotenv import load_dotenv

load_dotenv()
```

**Claude client input variables**

```python
claude_model="claude-haiku-4-5"
max_tokens_input=100
message_input=[
        {
            "role": "user",
            "content": "What is latest news about Hyderabad, India?    "
        }
    ]
tools_input = [{
    "type": "web_search_20250305",
    "name": "web_search",
    "max_uses": 3
}]
```

**Creating the Anthropic Client**

```python
from  anthropic import Anthropic

client = Anthropic()

print(f'''client : {client}, 
      \nCLAUDE_MODEL: {claude_model},
      \nmax_tokens: {max_tokens_input}
      \nmessage_input: {message_input},
      \ntools_input: {tools_input} 
    ''')
```

**Messages API with the Web Search Tool**

```python
response = client.messages.create(
    model=claude_model,
    max_tokens=max_tokens_input,
    messages=message_input,
    tools=tools_input
)
# print(f"Response: {response.content}")
```

**Claude web_search tool response payload**

```python
import json

print("ID:", response.id)
print("Role:", response.role)
print("Model:", response.model)
print("Stop reason:", response.stop_reason)
print("Usage:", response.usage)
print("usage input tokens:", response.usage.input_tokens)
print("usage output tokens:", response.usage.output_tokens)
print("server_tool_uses:", response.usage.server_tool_use)
# print("Title:", response.title)
# print("Url:", response.url)
# print("date:", response.page_age)
print("-*-"*20)
print("\nContent blocks:")
for i, block in enumerate(response.content):
    print(f"\n--- Block {i} ---")
    print("Type:", block.type)
    print(block.model_dump_json(indent=2))
    # if block.type == "text":
    #     print(json.dumps(block.model_dump(), indent=2))
    # if block.type == "web_search_result":
    #     print("Title:", block.title)
    #     print("Url:", block.url)
    #     print("Date:", block.page_age)
print("-*-"*20)
print("\nAs JSON:")
# print(json.dumps(response.model_dump(), indent=2))
print(response.model_dump_json(indent=2))
```
